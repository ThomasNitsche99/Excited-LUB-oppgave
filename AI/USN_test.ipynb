```python
from firecrawl_app import app, ExtractSchema, NestedModel1
import pprint
import json
import pandas as pd
from utils.helpers import *

filepath = "config.json"

# Load the JSON file containing config
with open(filepath, "r", encoding="utf-8") as json_file:
    config_data = json.load(json_file)
```

# Extracting LUB from UiB

```python



USN = config_data["USN"]

dfs = []
# Looping over study programs
for study_program, url in USN["study_programs"].items():
    
    print(f"Extracting data for {study_program} from {url} ..")
    
    data = app.extract([
            url #url to extract data from
            ], {
                'prompt': USN["prompt"],
                'schema': ExtractSchema.model_json_schema(),                
            })
    
    print("Data extracted ..")
    print("creating csv ..")
    LUBs = data["data"]["læringsutbyttebeskrivelser"]
    df = create_csv("USN", study_program, LUBs)
    dfs.append(df)    
    print("Scraping finished!")  
      
    
# Concatenating all dataframes
combined_df = pd.concat(dfs, ignore_index=True)
combined_df.to_csv(f"USN_LUBER.csv", index=False)
    
```
